```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
compas = pd.read_csv("../data/compas-scores-raw.csv")
```

```python
print('La distribution du sex')
print(compas['Sex_Code_Text'].value_counts())
print('-'*50)

print('La distribution des ethnies')
print(compas['Ethnic_Code_Text'].value_counts())
print('-'*50)

print('La distribution du langage')
print(compas['Language'].value_counts())
print('-'*50)

print('La distribution du statut marital')
print(compas['MaritalStatus'].value_counts())
print('-'*50)

print('La distribution des scores')
print(compas['RawScore'].value_counts())
print('-'*50)

print('La distribution du DecileScore')
print(compas['DecileScore'].value_counts())
print('-'*50)

print('La distribution du score')
print(compas['ScoreText'].value_counts())
print('-'*50)
```

```python
# Age
sns.histplot(compas['DateOfBirth'], kde=True)
plt.title('Age Distribution - COMPAS Recidivism')
```

```python
# Gender
sns.countplot(x='Sex_Code_Text', data=compas)
plt.title('Gender Distribution - COMPAS Recidivism')
```

```python
# Ethnicity 
sns.countplot(x='Ethnic_Code_Text', data=compas)
plt.title('Ethnicity Distribution - COMPAS Recidivism')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels by 45 degrees
plt.tight_layout()  # Adjust layout to prevent label cutoff
```
